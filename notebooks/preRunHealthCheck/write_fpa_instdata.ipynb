## Best Focus after adjusting cooler set Temperature in July/August 2024

```python
motors = {
    "r1" : [177.26, 134.17, 111.52],
    "r2" : [174.77, 203.92, 181.34],
    "r3" : [213.20, 204.29, 203.66],
    "r4" : [117.05, 69.89, 112.59],
    "b1" : [166.71, 114.53, 84.34],
    "b2" : [241.42, 259.41, 247.72],
    "b3" : [197.23, 187.09, 197.35],
    "b4" : [131.69, 94.08, 149.89],
}
```

```python
for cu, mot in motors.items():
    pos = mot - np.mean(mot)
    if cu[0] == "b":
        txt_cu = "  <<: *BLUE_XCU"
    if cu[0] == "r":
        txt_cu = "  <<: *RED_XCU"
    if cu[0] == "n":
        txt_cu = "  <<: *NIR_XCU"  
    print(f"{cu}:")
    print(txt_cu)
    print("  fpa:")
    print("    <<: *FPA")
    print(f"    tilts: [{pos[0]:.1f}, {pos[1]:.1f}, {pos[2]:.1f}]")
    print(f"    focus: {np.mean(mot):.1f}")
    print("\n")
```
